2. Modify the `calculate_total_price` function to include error handling.
   - If the user enters an invalid price (e.g., a negative value or a non-numeric value), display an error message and ask them to re-enter the price for that product.

3. Modify the `get_customer_orders` function to include error handling.
   - If the user enters an invalid number of orders (e.g., a negative value or a non-numeric value), display an error message and ask them to re-enter the number of orders.
   - If the user enters an invalid product name (e.g., a product name that is not in the inventory), or that doesn't have stock available, display an error message and ask them to re-enter the product name. *Hint: you will need to pass inventory as a parameter*
   - Use a try-except block to handle the error and continue prompting the user until a valid product name is entered.

4. Test your code by running the program and deliberately entering invalid quantities and product names. Make sure the error handling mechanism works as expected.


````python
# Define the products list
products = ["t-shirt", "mug", "hat", "book", "keychain"]

# 1. Initialize Inventory
def initialize_inventory(products):
    inventory = {}
    for product in products:
        while True:
            try:
                quantity = int(input(f"Enter the quantity of {product}s available: "))
            except ValueError:
                print("Error: Invalid input. Please enter a whole number.")
                continue
            if quantity < 0:
                print("Error: Quantity cannot be negative. Please enter a non-negative value.")
                continue
            inventory[product] = quantity
            break
    return inventory

# 2. Get Customer Orders
def get_customer_orders(inventory):
    # Ask for a valid number of orders
    while True:
        try:
            num_orders = int(input("Enter the number of customer orders: "))
        except ValueError:
            print("Error: Invalid input. Please enter a whole number.")
            continue
        if num_orders < 0:
            print("Error: Number of orders cannot be negative. Please enter a non-negative value.")
            continue
        break

    # Ask for a valid, in-stock product name for each order
    customer_orders = set()
    for _ in range(num_orders):
        while True:
            order = input("Enter the name of a product that a customer wants to order: ").strip().lower()
            try:
                if order not in inventory:
                    raise ValueError(f"'{order}' is not a valid product. Please choose from: {', '.join(inventory)}")
                if inventory[order] <= 0:
                    raise ValueError(f"'{order}' is out of stock. Please choose another product.")
            except ValueError as error:
                print(f"Error: {error}")
            else:
                customer_orders.add(order)
                break
    return customer_orders

# 3. Calculate Total Price
def calculate_total_price(customer_orders):
    total_price = 0
    for product in customer_orders:
        while True:
            try:
                price = float(input(f"Enter the price of {product}: "))
            except ValueError:
                print("Error: Invalid input. Please enter a number.")
                continue
            if price < 0:
                print("Error: Price cannot be negative. Please enter a non-negative value.")
                continue
            total_price += price
            break
    return total_price

# 4. Update Inventory
def update_inventory(customer_orders, inventory):
    inventory = {product: inventory[product] - (1 if product in customer_orders else 0) for product in inventory}
    inventory = {product: quantity for product, quantity in inventory.items() if quantity > 0}
    return inventory

# 5. Calculate Order Statistics
def calculate_order_statistics(customer_orders, products):
    total_products_ordered = len(customer_orders)
    percentage_ordered = (total_products_ordered / len(products)) * 100
    return (total_products_ordered, percentage_ordered)

# 6. Print Order Statistics
def print_order_statistics(order_statistics):
    print(f"""
Order Statistics:
Total Products Ordered: {order_statistics[0]}
Percentage of Unique Products Ordered: {order_statistics[1]}
    """)

# 7. Print Updated Inventory
def print_updated_inventory(inventory):
    print("Updated Inventory:")
    for product, quantity in inventory.items():
        print(f"{product}: {quantity}")

# 8. Main Program

# Step 1: Initialize the inventory
inventory = initialize_inventory(products)

# Step 2: Get customer orders
customer_orders = get_customer_orders(inventory)

# Step 3: Calculate order statistics
order_statistics = calculate_order_statistics(customer_orders, products)

# Step 4: Print order statistics
print_order_statistics(order_statistics)

# Step 5: Update the inventory
inventory = update_inventory(customer_orders, inventory)

# Step 6: Print the updated inventory
print_updated_inventory(inventory)

# Step 7: Calculate and print total price
total_price = calculate_total_price(customer_orders)
print(f"Total Price: {total_price}")
````

## Step 4: Testing the error handling

The cell below deliberately feeds the functions invalid input (non-numeric values, negative numbers, unknown product names and out-of-stock products) and checks that each one is rejected and re-prompted. The input is scripted, so it runs without typing anything. To try it by hand instead, run the main program cell above and type invalid values yourself.

````python
# Step 4: Test the error handling by feeding the functions deliberately invalid input
from unittest.mock import patch

def run_with_inputs(func, inputs, *args):
    """Run func while feeding it scripted inputs, echoing each prompt and answer."""
    answers = iter(inputs)
    def fake_input(prompt=""):
        answer = next(answers)
        print(f"{prompt}{answer}")
        return answer
    with patch("builtins.input", fake_input):
        return func(*args)

# Test 1: invalid quantities ("abc" = non-numeric, "-5" = negative), then valid ones
print("--- Test 1: initialize_inventory ---")
inv = run_with_inputs(initialize_inventory, ["abc", "-5", "10", "10", "0", "10", "10"], products)
assert inv == {"t-shirt": 10, "mug": 10, "hat": 0, "book": 10, "keychain": 10}

# Test 2: invalid number of orders, an unknown product, an out-of-stock product (hat has 0)
print("\n--- Test 2: get_customer_orders ---")
orders = run_with_inputs(get_customer_orders, ["x", "-1", "3", "sock", "hat", "mug", "book", "t-shirt"], inv)
assert orders == {"mug", "book", "t-shirt"}

# Test 3: invalid prices ("abc" = non-numeric, "-2" = negative), then a valid one
print("\n--- Test 3: calculate_total_price ---")
total = run_with_inputs(calculate_total_price, ["abc", "-2", "5"], {"mug"})
assert total == 5.0

print("\nAll tests passed: invalid input was rejected and the user was asked again.")
````
